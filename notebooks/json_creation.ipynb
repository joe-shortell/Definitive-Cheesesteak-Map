# Loading libraries

```python
import pandas as pd
import json
from pathlib import Path
import re
import unicodedata
```

# Creating JSON file

```python
# define the path to the CSV file
df = pd.read_csv("../data/locations.csv")

# Function to slugify location names
def slugify(value):
    value = str(value)
    value = unicodedata.normalize("NFKD", value).encode("ascii", "ignore").decode("ascii")  # remove accents
    value = re.sub(r"[^\w\s-]", "", value.lower())  # remove all non-word chars except whitespace and hyphens
    value = re.sub(r"[\s_-]+", "_", value).strip("_")  # collapse spaces/hyphens and remove leading/trailing underscores
    return value

# Create a list of dictionaries for each location
locations = []
for _, row in df.iterrows():
    slug = slugify(row["location"])
    entry = {
        "name": row["location"],
        "slug": slug,
        "description": row["description"] if pd.notna(row["description"]) else "No description available",
        "tier": int(row["tier"]),
        "latitude": row["latitude"],
        "longitude": row["longitude"]
    }
    locations.append(entry)

# Save JSON
Path("../data/locations.json").write_text(json.dumps(locations, indent=2), encoding="utf-8")
```
